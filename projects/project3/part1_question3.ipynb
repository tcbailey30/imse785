```python
import numpy as np

# Example 5.1 Matrix
M = np.array([
    [0, 0.5, 1.0, 0],
    [1/3, 0, 0, 0.5],
    [1/3, 0, 0, 0.5],
    [1/3, 0.5, 0, 0]
])

# Initial vector
v = np.array([0.25, 0.25, 0.25, 0.25])

print("Initial vector:", v)
print(v)
```

```python
# Number of iterations
num_iterations = 20

for i in range(num_iterations):
    v = M @ v  # Matrix-vector multiplication
    print(f"Iteration {i + 1}: {v}")
```
